```python
from google.colab import files
import pandas as pd

uploaded = files.upload()

train_file = next(f for f in uploaded if f.startswith("train_data") and f.endswith(".csv"))
test_file = next(f for f in uploaded if f.startswith("test_data") and f.endswith(".csv"))

train_data = pd.read_csv(train_file)
test_data = pd.read_csv(test_file)

print("Training Dataset:")
display(train_data.head())

print("Testing Dataset:")
display(test_data.head())

print("Train shape:", train_data.shape)
print("Test shape:", test_data.shape)
```

```python
train_data = train_data.drop(
    columns=["Unnamed: 0"], errors="ignore"
)

train_data = train_data.drop_duplicates()

train_data = train_data.dropna()

print("Cleaned Dataset:")
display(train_data)

print("Dataset Shape:", train_data.shape)
print("Missing Values:")
print(train_data.isnull().sum())
```

```python
age_in_years = train_data["Age"] * 80

result = train_data[
    (age_in_years > 10) &
    (train_data["Sex"] == 0)
]

print("Female Passengers Older Than 10 Years:")
display(result)
```

```python
class1 = train_data[train_data["Pclass_1"] == 1]

print("Passengers of Class 1:")
display(class1)

print("Statistical Description:")
display(class1.describe())
```

```python
train_data = train_data.drop_duplicates(
    subset=["PassengerId"]
)

print("Dataset after removing duplicate Passenger IDs:")
display(train_data)
```

```python
train_data.insert(1, "Category", "Passenger")

print("Dataset after inserting Category column:")
display(train_data.head())
```
